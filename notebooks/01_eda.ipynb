# MovieLens 25M — Keşifsel Veri Analizi

Bu notebook, ham CSV dosyaları üzerinde temel EDA yapar. Pipeline'a girmeden önce veri tanımak için.

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

DATA = Path('../data/ml-25m')

ratings = pd.read_csv(DATA / 'ratings.csv')
movies  = pd.read_csv(DATA / 'movies.csv')
tags    = pd.read_csv(DATA / 'tags.csv')
links   = pd.read_csv(DATA / 'links.csv')

print('ratings:', ratings.shape)
print('movies :', movies.shape)
print('tags   :', tags.shape)
print('links  :', links.shape)
```

```python
ratings.describe()
```

```python
ratings['rating'].value_counts().sort_index().plot(kind='bar', title='Rating dağılımı')
plt.show()
```

```python
user_counts = ratings.groupby('userId').size()
movie_counts = ratings.groupby('movieId').size()
print('Kullanıcı sayısı :', ratings.userId.nunique())
print('Film sayısı       :', ratings.movieId.nunique())
print('Aktif kullanıcı (>= 100 rating):', (user_counts >= 100).sum())
print('Popüler film (>= 1000 rating)  :', (movie_counts >= 1000).sum())
```

```python
ratings['year'] = pd.to_datetime(ratings['timestamp'], unit='s').dt.year
ratings.groupby('year').size().plot(title='Yıllara göre rating sayısı')
plt.show()
```

```python
movie_avg = ratings.groupby('movieId')['rating'].agg(['mean', 'count']).reset_index()
top = movie_avg[movie_avg['count'] >= 1000].nlargest(20, 'mean').merge(movies, on='movieId')
top[['title', 'mean', 'count', 'genres']]
```
